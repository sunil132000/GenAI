# Polymorphism

- In OOPs , Polymorphism allows objects of Different classes as if they are the object of the same class and It enables reusability and makes programs more flexible

### Types of Polymorphism:
    A. Method Overriding ( Runtime Polymorphism)
    B. Method Overloading ( compile Time Polymorphism)
    C. Operator Overloading



A. Method Overriding --> A child class provided a specific implementation of a method that has already been defined into the parent class , This allows Dynamic Selection at Runtime


```python
# Method Overriding
class Grandparent:
    def speak(self):
        print(f"hello my name is Amit")

class Parent(Grandparent):
    def speak(self):
        print(f"hello my name is Junior Amit")

class Child(Grandparent):
    def speak(self):
        print(f"hello my name is Junior Amit Junior")

def attribute_provider(object):
    return object.speak()


```

```python
grandpa = Grandparent()
attribute_provider(grandpa)
```

```python
parent = Parent()
attribute_provider(parent)
```

```python
child = Child()
attribute_provider(child)
```

```python
# Method Overloading
import math
class Shape:
    def area(self):
        return "This method returns the area of the shape"


class Rectangle(Shape):
    def __init__(self, length, width,height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return math.pi * self.radius * self.radius


shape1 = Shape()
print(shape1.area())

shape2 = Rectangle(10,20,30)
print(shape2.area())

shape3 = Circle(5)
print(shape3.area())
```

```python
# Operator Overloading
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value

    def __sub__(self, other):
        return self.value - other.value


n1 = Number(10)
n2 = Number(5)

print(n1 + n2)  # 15
print(n1 - n2)  # 5
```
