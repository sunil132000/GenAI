# Operator Overloading

- Operator Overloading allows us to define how operators (+,-,/,*,etc) work with user defines objects .
- why operator over loading userful ?
- Makes cutom objects behave like built in types (numbers , strings, lists)
- Improves code readability by using natural syntax by using natural syntax (Object1 + Object2) instead of object1.add(object2)

### Operator --> + --> Method name --> __add__(self , other)
### Operator --> - --> Method name --> __sub__(self,other)
### Operator --> * --> Method name --> __mul__(self,other)
### Operator --> / --> Method name --> __truediv__(self,other)
### Operator --> == --> Method name --> __eq__(self , other)
### Operator --> > --> Method name --> __gt__(self , other)

```python
class Avenger :
    def __init__(self,name,power):
        self.name = name
        self.power = power

    def __add__(self,other):
        return Avenger(self.name + "&" + other.name , self.power + other.power)

    # same way we can write other Method
```
