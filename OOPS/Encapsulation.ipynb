# Encapsulation

- is a way to wrapping data (variables and methods ) inside a capsule or a class as a Single Unit .
- It restricts the direct access to the objects component's which is a means of preventing accidental inference and misuse of the data.

### Key concepts of Encapsulation :
    A. Public Members - Accessible from Anywhere or Attributes are accessible outisde of the class and these members are accessible by the derived class . public member attribute is directly declared
    B. Protected Members -
    C. Private Members
    D. getter and setter methods

## Private Members --> A Private member is not accessible outside the class and not even in the derived class
and in order to have the access over these members or to bring some modification on these members we use getter and setter method .
- Private Members are declared by using double underscore __ before the name

```python
class Avenger:
    def __init__(self,name,phone_number):
        self.name = name
        self.__phone_number =phone_number

```

```python
aveng = Avenger("sunil","3523")
aveng.phone_number
```

```python
dir(aveng)
```

### getter Function

- It will be helping us out with accessing the private data

```python
class Captian:
    def __init__(self,name,address):
        self.name = name
        self.__address =address

    # getter method
    def accessing_add(self):
        print(f"My address is {self.__address}")

captain = Captian("Amit","New York")
captain.accessing_add()
```

### Setter Method --> Setter Method is being used for modifying our private data

```python
class Animal:
    def __init__(self,name,sound):
        self.name = name
        self.__sound =sound

    def sound_info(self):
        print(f"the Animal is going to sound as :{self.__sound}")

    def sound_modify(self,new_sound):
        self.__sound = new_sound

animal = Animal("sunil","meo")
animal.sound_info()
animal.sound_modify("bho bho")
animal.sound_info()

```

### Protected Data

- is the Data which is not accessible outside the class But the protected data is going to be accessible by the derived class or child class
- for having the protected member you can mention your protected member as __attribute or by using a single underscore


```python
class GrandParent:
    def __init__(self,name,locker_key):
        self.name = name
        self._locker_key = locker_key

    # getter method
    def locker_info(self):
        print(f"The locker key is {self._locker_key}")


class Parent(GrandParent):
    def __init__(self,name,locker_key):
        super().__init__(name,locker_key)

    def child_reveals(self):
        print(f"The Locker Key information is: {self._locker_key}")


parent = Parent("Sunil",123)
parent.child_reveals()
```
