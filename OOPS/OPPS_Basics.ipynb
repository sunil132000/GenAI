#### Types of Methods
 A. Instance Method -  Instance Method Operate on Object Attributes , his Instance Method is goin to use self keyword to access and modify object attributes . these Methods are called using an object .<br>

 B. static Method - static Methods do not use self or cls , They work like normal functions but belong to class only, They are gon to be designed by a abbreviation @staticmethod , static method can also be called by the classname , It’s just a utility function grouped inside a class.<br>

 C. Class Method - It is the method of whole class or the whole blueprint which will remain the same for all the objects then such kind of Method is being reffered as a class Method . In Class Method , these methods work with the class level attributes , they are going to use "cls" instead of self , they are called using class name as well , For specific that the current method is a class method we have to use the abbreviation as @classmethod <br>

```python
class Avengers:
    #class attribute
    origin ="Avenger"

    def __init__(self,name,power):
        self.name = name
        self.power = power

    # class Method
    @classmethod
    def origin_info(cls):
        print(f"My origin is {cls.origin}")

    @staticmethod
    def mission():
        print(f"All Avenger must protect the Earth ")

    #Instance Method
    def self_info(self):
        print(f"My self is {self.name}")


Iron_man=Avengers("Iron Man","suit")
Iron_man.power
Iron_man.self_info()
Iron_man.origin_info()
Iron_man.mission()
```
