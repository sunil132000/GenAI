# Pillars of Object Oriented Programming

### Inheritance - is a Fundamental concept in Object Oriented Programming because of which the child class is able to inherit the properties OR Attributes and functionalities of the parent

#### Types of Inheritance
     A. Single Inheritance<br>
     B. Multilevel Inheritance<br>
     C. Multiple Inheritance<br>

```python
# Single Inheritance  --> The Feature or the Attributes from one parent class will be getting inherited into one child class

class Avenger:
    def __int__(self,name,power):
        self.name = name
        self.power = power

    def introduce(self):
        print(self.name,self.power)

class Ironman(Avenger):
    def __init__(self,name,power,weapon):
        super().__int__(name,power)
        self.weapon=weapon

    def weapon_detail(self):
        print(f"I use  {self.weapon}")

iron = Ironman("Ironman","Genius Intellect","leaser beam")
iron.introduce()
iron.weapon_detail()
```

```python
# Multilevel Inheritance  --> The Properties or Methods and Attributes of the grandparent class can be inherited by the parent class and then the properties of the parent class will be inherited by the child class . In proper heirarchy the Inheritance process is going to work for us .


class Grandparent:
    def __init__(self,nationality,baldness,anger_issues):
        self.nationality = nationality
        self.baldness = baldness
        self.anger_issues = anger_issues

    def introduce(self):
        print(f"Hello my nationality is :{self.nationality} and about my anger issues if you want to ask about then answer is :{self.anger_issues} ")

class Parent(Grandparent):
    def __init__(self,nationality,baldness,anger_issues,intelligence):
        super().__init__(nationality,baldness,anger_issues)
        self.intelligence=intelligence

    def intelligence_detail(self):
        print(f"the intelligence is {self.intelligence}")

class child(Parent):
    def __init__(self,nationality,baldness,anger_issues,intelligence,charm):
        super().__init__(nationality, baldness, anger_issues,intelligence)
        self.charm = charm

    def char_details(self):
        print(f" About my charm if you want to know {self.charm}")


# object creation
parent_object = Parent("Indian","Yes","No","Yes")
parent_object.introduce()
child_object = child("Indian","less","not anger","intelligent","charm")
child_object.introduce()

```

- The diamond problem in Python occurs in multiple inheritance when a class inherits from two classes that have a common parent, creating a diamond-shaped structure. This can cause confusion about which method should be executed. However, Python resolves this problem using Method Resolution Order (MRO), which follows the C3 linearization algorithm. MRO ensures that methods are searched in a fixed order: first the child class, then parent classes from left to right, and then the common ancestor, without repeating any class.

```python
# Multiple Inheritance --> In Multiple Inheritance , The properties of two parent class or multiple parent class will be getting inherited into the child class .

class Ironmans:
    def __init__(self,name,power,suit,weapon,nationality="American"):
        self.name = name
        self.power = power
        self.suit = suit
        self.weapon =weapon
        self.nationality = nationality

    def introduction(self):
        print(f"My name is {self.name} and power is :{self.power} ")

class Captainamerica:
    def __init__(self,name,power,suit,weapon,nationality="American"):
        self.name = name
        self.power = power
        self.suit = suit
        self.weapon =weapon
        self.nationality = nationality
    def weapon_detail(self):
        print(f"my weapon is {self.weapon}")

class Hulk(Ironmans , Captainamerica):

    def __init__(self,name,power,suit,weapon,nationality,anger_issues):
        super().__init__(name,power,suit,weapon,nationality)
        self.anger_issues =anger_issues

    def anger_description(self):
        print(f"About my anger issues if i am having or not then the answer is :{self.anger_issues}")


hulk_guy = Hulk("hulk","Amazing Power and Strength","Mark 85 Ironman Suit","Repulsor beams and Shield","American","Yes")

hulk_guy.introduction()
hulk_guy.weapon_detail()



```
