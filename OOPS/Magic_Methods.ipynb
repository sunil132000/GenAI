# Magic Methods

- Magic Methods are also called as Dunder Methods because they have double underscores __ before and after their name and they are special methods in python that allow us to customize how objects behave

### These Methods can help us define
     A.Object Intialization (__init__)
     B. String Representation (str,repr)
     C. Arthematic Operatiosn (add, sub,etc)
     D. Comparison Operations (eq,It)
     E. Length , Indexing , Iteration (len,getitem,etc)

### Why to use Magic Methods
    A. Makes Objects behave like built in types (eg, lists, integers)
    B. Improves Readability by allowing basic operations objects1 + objects2
    C. Simplifies code by repalcing verbose function calles with natural syntax

```python
# print(objects)  --> __str__ --> what i returned when print(object) is used

# Arithematic operations can be done among the objects by using magic methods
# __add__

class Avenger:
    def __init__(self, name, strength):
        self.name = name
        self.strength = strength

    def __add__(self, other):
        return Avenger(self.name + "&" + other.name,
                       self.strength + other.strength)

    def __str__(self):
        return f"{self.name} with strength {self.strength}"


ironman = Avenger("Ironman", 80)
thor = Avenger("Thor", 49)

team = ironman + thor
print(team)
```
