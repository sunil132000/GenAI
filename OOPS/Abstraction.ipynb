# Abstraction

- is the Concept of Hiding the implementation details and only showing the necessary features of the object .
- This helps in reducing programming complexity and efforts
- Abstraction allows us to work with an object with what an object does rather than how it does it
- ABC (from abc) is used to create abstract classes with methods defined using abstractmethod.
Yes, any subclass must implement all abstract methods, otherwise its object cannot be created.

### How can we achieve abstraction in python
    A. Abstact class  -  A class that cannot be instantiated or that means for an abstract class we cant create the object of this class

    B. Abstract method - Abstract method just because it cannt accessed by the object of the abstract class itself thats why we have to call this abstract method in our child class and the implentation of the abstract method is not provided iside the abstract class but must be provided in the child class

    C. The abstract method is being dfined by @abstactmethod


```python
from abc import ABC, abstractmethod

class Vechile(ABC):

    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass


class Car(Vechile):
    def start(self):
        print(f"The car starts with an ignition key")

    def stop(self):
        print(f"The car stops with an breaks pedals")


car = Car()
car.start()
car.stop()
```
