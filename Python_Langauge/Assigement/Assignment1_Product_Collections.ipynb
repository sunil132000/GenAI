# Product Collections(List and Tuples)

```python
# Creating a list of 6 product names
product_name = ["watch", "tv", "mobile", "laptop", "bag", "bottle"]

# Creating a tuple with (product_name, price, category)
sample_product = ("watch", 1000, "Electronics")

# Printing the 2nd product (index 1)
print(product_name[1])

# Printing the last product using length
print(product_name[len(product_name) - 1])

# Appending two new products to the list
product_name.append("cloths")
product_name.append("wires")

# Printing updated product list
print(product_name)


# ---------------- Optional Part ---------------- #

# Converting tuple to list (so we can modify it)
converted_tuple_to_list = list(sample_product)

# Printing converted list
print(converted_tuple_to_list)

# Checking type (should be list)
print(type(converted_tuple_to_list))

# Changing the price (index 1)
converted_tuple_to_list[1] = 2000

# Printing updated list
print(converted_tuple_to_list)

# Converting list back to tuple
list_to_tuple = tuple(converted_tuple_to_list)

# Checking type (should be tuple)
print(type(list_to_tuple))

# Printing final tuple
print(list_to_tuple)
```

# Product Pricing(Dictionaries)

```python
# Task 3: Product Pricing (Dictionaries)

# 1. Create price_dict with at least 6 entries
price_dict = {
    "Apple": 1.50,
    "Banana": 0.75,
    "Mango": 2.00,
    "Orange": 1.20,
    "Grapes": 3.50,
    "Watermelon": 5.00
}

print("Original price_dict:", price_dict)

# 2a. Add a new product
price_dict["Pineapple"] = 4.00
print("\nAfter adding Pineapple:", price_dict)

# 2b. Update the price of an existing product
price_dict["Apple"] = 2.00
print("\nAfter updating Apple price:", price_dict)

# 2c. Remove a product by name (handle case when product does not exist)
product_to_remove = "Banana"
if product_to_remove in price_dict:
    price_dict.pop(product_to_remove)
    print(f"\n'{product_to_remove}' removed successfully.")
else:
    print(f"\n'{product_to_remove}' does not exist in the dictionary.")

print("After removal:", price_dict)

# Try removing a product that doesn't exist
product_to_remove = "Strawberry"
if product_to_remove in price_dict:
     price_dict.pop(product_to_remove)
else:
    print(f"\n'{product_to_remove}' does not exist in the dictionary.")

# 3. Print average price
total = sum(price_dict.values())
count = len(price_dict)
average = total / count
print(f"\nAverage price of all products: {average:.2f}")

# Extra: Product with maximum and minimum price
max_product = max(price_dict, key=price_dict.get)
min_product = min(price_dict, key=price_dict.get)
print(f"\nMost expensive product : {max_product} -> ${price_dict[max_product]:.2f}")
print(f"Cheapest product       : {min_product} -> ${price_dict[min_product]:.2f}")
```

# Combined Operations

```python
# List of products along with their categories
products = [
    ("Apple", "Fruits"),
    ("Mango", "Fruits"),
    ("Orange", "Fruits"),
    ("Grapes", "Fruits"),
    ("Watermelon", "Vegetables"),
    ("Pineapple", "Vegetables")
]

# Price list of products (some items might be missing intentionally)
price_dict = {
    "Apple": 1.50,
    "Banana": 0.75,
    "Mango": 2.00,
    "Orange": 1.20,
    "Grapes": 3.50,
    "Watermelon": 5.00
}

# This will store final product details (name, price, category)
catalog = []

# Loop through each product and check if we have its price
for product_name, category in products:

    # Only include product if price is available
    if product_name in price_dict:
        price = price_dict[product_name]

        # Add complete product info to catalog
        catalog.append((product_name, price, category))

# Let’s see what our catalog looks like
print(catalog)


# Now we want to group products by category
category_to_products = {}

# Loop through catalog and organize products category-wise
for product_name, price, category in catalog:

    # If category not already present, create a new list
    if category not in category_to_products:
        category_to_products[category] = []

    # Add product to its category
    category_to_products[category].append(product_name)

# Print category-wise products in a clean format
print("\nCategory to Products:")
for category, product_list in category_to_products.items():
    print(f"  {category}: {product_list}")


# Find which category has the most products
largest_category = max(category_to_products, key=lambda x:len(category_to_products[x]))

# Print the category with maximum products
print(largest_category)
```
