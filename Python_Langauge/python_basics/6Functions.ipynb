## Recursive Function

-By a Recursive Function we can say that a Function is going to call itself again and again

```python
# Factorial of a Number
def Factoiral(number):
    if number==0 or number==1 :
        return 1
    else:
        return number * Factoiral(number-1)
```

```python
Factoiral(5)
```

### Function with Variable length Arguments

- Positional Arguments --> Position arguments are being used so that we can pass multiple values inside our function as an a argument irrespective of number of values we have mentioned as the parameter inside our function

```python
## function to add multiple  numbers irrespective of how many numbers we are passing

def addition_positional(*num):
    add =0
    for i in num :
        add+=i
    print("the addition of the numbers are" ,add)
```

```python
addition_positional(10,2,3,4,5,6,7,8,1000)
```

### Keyword Arguments --> Takes the input as the keyword values and irrespective of any number of keyword values you are going to pass it wont matter

```python
def data(**user_data):
    for key , value in user_data.items():
        print(f"key:{key} value:{value}")
```

```python
data(name="Amit",age=25,gender="male")
```

### you guys wnat to pass keyword value pairs as your input along with positional arguments as well

```python
def variable(*positional,**keyword):
    for i in positional :
        print(f'positional:{i} keyword:{keyword}')
    for key,vakue in keyword.items():
        print(f'key:{key} value:{vakue}')
```

```python
variable(1,2,3,4,5,name="Amit",Age=25)
```
