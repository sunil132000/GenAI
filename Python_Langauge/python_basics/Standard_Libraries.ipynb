- csv library --> MOst of the time we are going to work with csv file so thats why knowledge with csv library functionality matters

```python
import csv

with open('example.txt', 'w', newline='') as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(['Name','Age','Gender'])
    writer.writerow(['','19','Male'])
```

```python
with open('example.csv', 'w', newline='') as csvfile:
    reader = csv.reader(csvfile)
    for row in reader:
        print(row)
```

- Datetime Library ---> It helps us out with figuring out the current time and date

```python
from datetime import datetime,timedelta

time_Current = datetime.now()
```

```python
# timedelta --> it give difference between the current time to provided time

yesterdate_time = time_Current - timedelta(days=1)
print(yesterdate_time)
```

- time --> time library provided us the current time and if you want your interpreter to sleep for seconds you can do it by using time library
- If you are writing a piece of code and you want your interpreter to sleep for a number of seconds then you can make the use of sleep function

```python
import time
print(time.time())
time.sleep(4)
print(time.time())
```

-Json Libray- when we hoing to work with Nosql Database data then the data Inside the Nosql Database is going to be their in the json format and along with that when we are going to work with the datasets as well then again these dataset will be consisting of data in the json format itself

```python
# when we get the Json data , we get in a json string format so by using json library we can convert our string data into dictionary format for serializing our data

import json

data = {"Name":"Amit","Age":25}
json_str = json.dumps(data) # dumps() --> json to str
print(json_str)
print(type(json_str))
```

```python
# json.loads()--> str to json
json_dict = json.loads(json_str)
print(json_dict)
print(type(json_dict))
```

- Regular Expression - is used to retrieve info based on a specific pattern or if the pattern matches

```python
import re

text = 'Hello My name is Amit and my password is HelloAmit12345'

pattern =r'\d+'
matching = re.search(pattern,text)
print(matching.group())
```
