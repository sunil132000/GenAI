# While Loop

- while loop is being used for iterating over a series of numbers when a condition is true and we have to intilialise a values as well iterating

```python
# by making the use of while loop we need to print the numbers from 0 to 50
i=0
while i<=50:
    print(i)
    i= i+1

```

```python
# we have to print the numbers from 0 to 50 as the even numbers

i=0
while i<=50:
    if i%2==0:
        print(i)
    else :
        pass
    i=i+1
```

### Loop Control Statements

- Loop Control statments are being used for terminating a loop in between wehn a condition is met or if you want to have some control over your iterating process ro while iterating over the series of numbers
- break --> break is our first loop control statement which help you out to terminate the loop when a certain condition is met
- continue ---> continue is our second loop control statement which helps us in iterating over a series of numbers but helps us with skipping the values in between once we meet a condition
- Pass ---> Pass statement is being used for doing nothing


### Nested Loops

#### f string --> f string will be helping us out with printing the numbers with the variable value present inside the colons so we can directly represent the value inside the colons message itself where variable value will be shown inside the {}

```python
# Nested loops can help us out iterating over a series of numbers as with the proper combination of iterating numbers

for i in range(6):
    for j in range(4):
        print(f" i value is :{i}  j value is :{j}" )
```

```python
# print the sum of first N natural numbers
n_value = int(input("enter a number"))
sum =0
for i in range(n_value+1):
    sum +=i
print(sum)
```

```python
# print the prime number from 0 to 100

for i in range(2,101):
    for j in range(2,i):
        if i%j==0:
            break
        print(i)
```
