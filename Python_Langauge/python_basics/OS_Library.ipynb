# OS Library

- OS Library is able to interact with our Operating System and can let us work with Multiple Directories and Files

#### getcwd() --> It is capable of providing us with the idea the current working directory

- getcwd  --> get current working directory

```python
import os
print(os.name)
```

```python
current_path = os.getcwd()
print(current_path)
```

#### mkdir()  --> make directory   ---> It will be helping s out with creating the directory

```python
new_path ="New_Folder"
os.mkdir(new_path)
print(f"The new directory is {new_path} is created successfully")
```

#### listdir() --> Listing all of the directories present

```python
print(os.listdir('.')) # . represent from present folder
```
