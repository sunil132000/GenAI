# Exception Handling

- Exception Handling is a way in which if we are getting the Errors in programming or if the user entered input causes an error , then instead of Throwing an Error we can rise an exception which can let the user know that the input is Not Correct or the given input can come up as an Error. As we donot want to stop the execution of the Program , so we raise an error which will let the user know that the current input is Not correct but the Program Execution wont stop .
- Exception are events that disrupt the normal flow of a program .They occur when an error is encounterd during program execution.
