# Lambda Function

- Lambda Function are the One Liner Functions defined by the Lambda Keyword
- Syntax<br>
   lambda arguments: expression<br>
   lambda --> keyword which simpally le the pyton interpreter know that the current function is our Lambda Function<br>
    Arguments --> Input values for our Function
    Expression --> Logic (Function Body )<br>
- They can have any number of Arguments but only one expression or only one Logic
- They are used for short operations or as Arguments to Higher Order Functions
- It can be treated as a Function without a Name and they Donot even require a Return Statement

```python
#  calculate the sqaure of a number

a = lambda x :x**2
type(a)
```

```python
a(2)
```

```python
 # Lambda function to calculate the sum of two numbers
 sum = lambda x,y:x+y
 sum(2,4)
```
