## File Handling Code

- Read the content from My File in the system
  1. open() --> opening your file
  2. First parameter --> File_name
  3. Second Parameter --> Mode ( Mode--> It decides the way in wich we are working with this file
  4. Mode --> r --> Reading the content from the file
  5. Mode --> w --> Write the content into the file
  6. Mode --> r+ --> Read the content first and then writing the content into the file
  7. Mode --> w+ --> Writing the content first and then reading form it
  8. Mdde --> a --> appending the data from the last line of the file

#### Reading your File
-read() --> Reads the content from the file

```python
f =open('Utility/example','r')
content = f.read()
print(content)
f.close()
```

### Wrtiting Methods into the File


- write() --> helps us out in writing the content into the File( it overwrites the content into the text file means firstly it erase all the content and then it write )

- append() --> it adds the data from the end of the file

Note --> If you do not close the file it do not write into the file


```python
#write() remove all content of file then write into it
f = open('Utility/example','w')
content = f.write("helasfdasdflo")
f.close()
```

```python
#append() it append data from the last line of the file
f = open('Utility/example','a')
content = f.write(" sunil")
f.close()
```

### with --> IT will be used for first of all opening the file as a object and there is not requirement for closing the file as well

```python
with open('Utility/example','r') as f:
    content_data = f.read()
    print(content_data)
```

```python
# Write the content into our file line by line  --> writelines()
line = ['First_line\n','Second_line\n','Third_line\n']
with open('Utility/example','w') as f:
    f.writelines(line)
```

### Mode --> Write the content into the binary file  --> wb
### Mode --> Read the content from the binary file  --> rb

```python
data =b'\x00\x01\x02\x03\x04'
with open('Utility/example','wb') as f:
    f.write(data)
```

#### Reading the Total Lines , Total words , Total Characters from a file  --> readlines()

- readlines() --> Reads the line in a way where in it is like the lines are stored inside a list

```python

def count_lines(file):
    with open(file,'r') as f:
        lines = f.readline()
        total_lines = len(lines)
        total_words =sum(len(x.split()) for x in lines)
        total_character = sum(len(line) for line in lines)
    return total_lines,total_words,total_character
```

- w+ --> first of all as W is their so it states that first writing operation will be done and then Reading operation will be done
- seek() --> It is going to help us out with reading the content after writing from starting only

```python
# example
with open('Tota_data.txt','w+') as file:
    file.write("\n we are looking into W+ operation")
    file.seek(0) # as previoudly it write so pointer is at last to seek(0) it come to first so that we read from starting
    f= file.read()
    print(f)
```
