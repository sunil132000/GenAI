# Map Function

-Map function applies a given function operation to All items in an iterable or a List return a Map object(An Iterator)

- map function is particularly useful for Transforming the data or applying on operation on ALl items of the iterable
- map() is going to help us out avoiding unnecessary lines of code and we can apply the operation on all values or all items in an iterable and this is the sign of a good programmer
- Syntax:<br>
   list(map(function,iterable))

```python
# Square all items of a list
a =[1,2,3,4,5,6,7]
b=list(map(lambda  x:x**2,a))
print(b)
```

#### Whenever you are going to pass the general function inside your map function as the parameter then you do not have to use parenthesis<br>
ex-print(list(map(squared_function ,[1,2,3,4,5])))

### While Mapping the value of one iterable with another iterable we can actually apply the operation in between the value of the two iterable in a synchronous way

[1,2,3,4]  <-->[5,6,7,8]  adding both list---->[6,8,11,6]

```python
# Map Function with Multiple Iterable
number_1 = [1,4,5,6]
number_2 = [4,7,2,9]
print(list(map(lambda x,y:x+y,number_1,number_2)))
```

```python
# Map() TO convert all string values items in an iterable into an integer

print(list(map(int,['1','2','3','4','5','6'])))
```

```python
# convert each string in all case
a=['amit','sunil','singh']
print(list(map(str.upper,a)))
```

```python
# Retrieve the values inside a dictionary using Map()
employess = [{'Name':'Amit','Age':23,'Qualification':'Btech'}
    ,{'Name':'sunil','Age':'25','Qualification':'Btech'}]

def age_retiever(list_item):
    return list_item['Age']

print(list(map(age_retiever,employess)))
```
