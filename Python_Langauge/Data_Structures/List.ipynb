## List Indexing

```python
# Reading the values from list
# List Indexing - By using the proper indexing value we can retrieve the data from our lists
#Indexing in List ---> 0 to length-1
lists =[ "Amit" , 1,3,5,6.7, True]
print(lists[5])
#len() method is used to get the length of the list
print(len(lists))
```

```python
#Negative Indexing ---> In any data structure we can access the data using Negative indexing as well ( getting value from back of the list)
lists[-2]
```

```python
# For having the access for all the values you are going to store inside the list , you can use ":"
print(lists[:]) # Copy of list , New object
print(lists) # Original list ,Same object
```

## List Slicing

## List slicing can helps as out with retrieving a specific data points from out list


```python
lists = ['mango', 'banana ', 'apple', 'orange', 'watermelon ', 'gavava' , 'grapes']
lists[1:4] # last index is not included
```

```python
# Reverse Order
lists[::-1]
```

# List are Mutable

## By mentioning that Lists are mutable we mean to say that we can replace the values present at a specific location inside our list

```python
lists[2] = "sunil"
print(lists)
```

### By using Looping Statements we can iterate through List as well

```python
for item in lists:
    print(item)
```
