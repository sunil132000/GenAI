# Tuples Method


## tuples , we are going to have 2 method usually as:

  ### 2.index()


```python
tuple_method = (1,2,3,4,5,7,8,9,2,4,3,2,6,78)
b= tuple_method.index(3)
a= tuple_method.count(2)
print(a ,b )
```

### Packing and unpacking of Tuples

```python
# Packing of tuples
# if you store multiple value in a single variable then default it store inside the tuple
a= 1,2,3,4,5,6,45,21,78,17
type(a)
```

```python
#UnPacking of tuples
#storing each value of tuple in a variable
a,b,c,d,e = 1,2,3,6,"sdfs"
print(a,b,c,d,e)
```

```python
## unpacking with a *
# in this I want to store first value in first variable and last in last variable and rest value in middle variable
first_value , *b , final_value = (1,2,3,4,5,6,67,8,9)
```

```python
print(final_value,b,final_value)
```
